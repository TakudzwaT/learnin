```python
New_X_train = X_train.copy()
for feature in range(0,46):
    mean = New_X_train[feature].mean()
    for idx, row in New_X_train.iterrows():
        if pd.isnull(row[feature]):
            New_X_train.at[idx, feature] = mean
```
